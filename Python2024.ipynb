Nu tittar vi på den andra formen av loop, while loop. Den körs så länge ett givet vi villkor är uppfyllt.

## 3.8 Egendefinierade funktioner

Man kan definiera egna funktioner med hjälp av kommandot **def**. 

```python
def apa(x): #Här anges båda att funktionen skall heta Apa och att den skall ta en invärde x
    return 4*x**2+4*x+1 #Här anges det som funktionen skall returnera, dvs funktionsvärdet
```

Man kan sedan anropa funktionen.

```python
apa(5)
```

Funtionsvärdet blir i detta fall ett tal som man kan använda i beräkningar. Kan du förutsäga vad nedanstående kommer att ge?

```python
apa(5)+apa(6)+2
```

```python
apa(apa(5))
```
